```python
# Se guarda en una nueva lista el nombre de jugador
names = jugadores_df.Jugador.tolist()
```

```python
# Eliminamos el campo de jugador por tipo de varible para poder realizar el escalado
jugadores_df = jugadores_df.drop(['Jugador','categoria'], axis = 1)
```

```python
# Creamos una matriz de correlacción
import seaborn as sns

corr = jugadores_df.corr()
ax = sns.heatmap(
    corr,
    vmin=-1, vmax=1, center=0,
    cmap=sns.diverging_palette(20, 220, n=200),
    square=True
)
ax.set_xticklabels(
    ax.get_xticklabels(),
    rotation=90,
    horizontalalignment='right'
);
```

```python
# Actividad 4 Métodos de aprendizaje No supervisado
# Analisis de datos torneo de futbol

# Escalamos los datos de los campos de la DB, para poder realizar comparacion

from sklearn import preprocessing
x = jugadores_df.values # numpy array
scaler = preprocessing.MinMaxScaler()
x_scaled = scaler.fit_transform(x)
X_norm = pd.DataFrame(x_scaled)
```

```python
# Mostramos los campos escalizados
X_norm.head()
```

```python
# Visualizamos los datos
jugadores_df.head()
```

```python
# Graficamos los datos escalizados
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns

fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(10, 6))

ax1.set_facecolor('#E8E8F1')
ax2.set_facecolor('#E8E8F1')

ax1.set_title('Antes de la transformacion')
sns.kdeplot(jugadores_df['Minutos_jugados'], ax=ax1)
sns.kdeplot(jugadores_df['Goles'], ax=ax1)
sns.kdeplot(jugadores_df['Tiros'], ax=ax1)
ax2.set_title('Despues de la transformacion')
sns.kdeplot(X_norm[1], ax=ax2)
sns.kdeplot(X_norm[2], ax=ax2)
sns.kdeplot(X_norm[8], ax=ax2)
plt.show()
```

```python


# Operación de cluterización con el algoritmo k-means
from sklearn.decomposition import PCA
pca = PCA(n_components = 2) # 2D PCA for the plot
reduced = pd.DataFrame(pca.fit_transform(X_norm))
```

```python
from sklearn.cluster import KMeans
# specify the number of clusters
kmeans = KMeans(n_clusters=4)
# fit the input data
kmeans = kmeans.fit(reduced)
# get the cluster labels
labels = kmeans.predict(reduced)
# centroid values
centroid = kmeans.cluster_centers_
# cluster values
clusters = kmeans.labels_.tolist()
```

```python
reduced['cluster'] = clusters
reduced['name'] = names
reduced.columns = ['x', 'y', 'cluster', 'name']
reduced.head()
```

```python
print(reduced.groupby('cluster').size())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set(style="white")
ax = sns.lmplot(x="x", y="y", hue='cluster', data = reduced, legend=False,
fit_reg=False, height = 15, scatter_kws={"s": 250})
texts = []
for x, y, s in zip(reduced.x, reduced.y, reduced.name):
    texts.append(plt.text(x, y, s))
    ax.set(ylim=(-2, 2))
plt.tick_params(labelsize=15)
plt.xlabel("PC 1", fontsize = 20)
plt.ylabel("PC 2", fontsize = 20)
plt.show()
```
